```python
import numpy as np
import matplotlib.pyplot as plt
```

# Exercise 1
plot a sine wave with frequency $f=5$Hz and amplitude $2.2V$ from $t=0$s until $t=5$s with N=100 points.

```python
t = np.linspace(0, 5, 100)
v = 2.2*np.sin(5 / 2 / np.pi * t)

plt.plot(t, v)
plt.xlabel("t (s)")
plt.ylabel("v (V)")
plt.show()
```
