```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 20, 100) # test
d = 1.2
y = np.cos(x + d)

plt.plot(x, y)
plt.plot(x, np.arctan(x))
plt.xlabel('some x label')
plt.ylabel('some y label')
plt.grid()
plt.show()
```

```python
x = np.linspace(3, 20, 50) # test
d = 1.2
y = np.tanh(x + d)

plt.plot(x, y)
plt.xlabel('some t label')
plt.ylabel('some v label')
plt.grid()
plt.show()
```
